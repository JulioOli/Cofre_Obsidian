# Rateio de Gastos das Prensas - Março

Este notebook inicia a análise de rateio dos gastos das prensas da empresa no mês de março.

Nesta primeira etapa, vamos carregar os dados de referência do arquivo `Prensas.csv` para um DataFrame e preparar a base para os próximos passos do rateio.

```python
from pathlib import Path
import pandas as pd

# Tenta localizar o CSV em cenários comuns de execução do notebook.
possiveis_caminhos = [
    Path.cwd() / "02-Referencias" / "Meus_Dados" / "Prensas.csv",       # quando cwd = raiz do projeto
    Path.cwd().parent / "02-Referencias" / "Meus_Dados" / "Prensas.csv",  # quando cwd = 04-Notebooks
]

caminho_csv = next((p for p in possiveis_caminhos if p.exists()), None)

if caminho_csv is None:
    caminhos_texto = "\n".join(str(p) for p in possiveis_caminhos)
    raise FileNotFoundError(f"Não foi possível localizar o arquivo Prensas.csv.\nTentativas:\n{caminhos_texto}")

# CSVs de sistemas brasileiros geralmente vêm em cp1252/latin1 e separador ';'.
ultimo_erro = None
for enc in ["utf-8", "utf-8-sig", "cp1252", "latin1"]:
    try:
        df_prensas = pd.read_csv(caminho_csv, sep=None, engine="python", encoding=enc)
        encoding_usado = enc
        break
    except UnicodeDecodeError as e:
        ultimo_erro = e
else:
    raise ultimo_erro

print(f"Arquivo carregado: {caminho_csv}")
print(f"Encoding detectado: {encoding_usado}")

df_prensas.head()
```

```python

# Validação rápida do DataFrame carregado
print("Dimensões do DataFrame:", df_prensas.shape)
print("\nTipos de dados por coluna:")
print(df_prensas.dtypes)
print("\nQuantidade de valores nulos por coluna:")
print(df_prensas.isnull().sum())

```

```python
# Converte colunas com "valor" no nome para float
colunas_valor = [col for col in df_prensas.columns if "valor" in col.lower()]
df_prensas[colunas_valor] = df_prensas[colunas_valor].replace(",", ".", regex=True).apply(pd.to_numeric, errors='coerce')
print(df_prensas.dtypes)
```
